# VideoGen — Colab GPU preflight
This only inspects available hardware; **no model is downloaded and no generation has been run**.
Enable a GPU runtime in Colab first. Read docs/operations/GPU_COLAB.md before downloading checkpoints.


```python
import os, platform, shutil, subprocess, json
report = {'python': platform.python_version(), 'disk_free_gib': round(shutil.disk_usage('/content').free/2**30, 2)}
try:
    import torch
    report['torch'] = torch.__version__
    report['cuda_available'] = bool(torch.cuda.is_available())
    if torch.cuda.is_available():
        p = torch.cuda.get_device_properties(0)
        report['gpu'] = p.name
        report['vram_gib'] = round(p.total_memory/2**30, 2)
        report['cuda_version'] = torch.version.cuda
except Exception as e:
    report['torch_error'] = str(e)
try:
    report['nvidia_smi'] = subprocess.run(['nvidia-smi', '--query-gpu=name,memory.total', '--format=csv,noheader'], capture_output=True, text=True, timeout=8).stdout.strip()
except Exception as e:
    report['smi_error'] = str(e)
print(json.dumps(report, indent=2))
```
